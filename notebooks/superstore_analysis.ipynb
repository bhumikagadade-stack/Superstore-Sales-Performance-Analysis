```python
query = """
WITH product_profit AS (
    SELECT
        "Product ID",
        "Product Name",
        SUM(Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM superstore
    GROUP BY "Product ID", "Product Name"
)

SELECT
    "Product ID",
    "Product Name",
    Total_Sales,
    Total_Profit,
    RANK() OVER (
        ORDER BY Total_Profit DESC
    ) AS Profit_Rank
FROM product_profit
ORDER BY Profit_Rank
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH product_profit AS (
    SELECT
        Category,
        "Product ID",
        "Product Name",
        SUM(Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM superstore
    GROUP BY
        Category,
        "Product ID",
        "Product Name"
),

ranked_products AS (
    SELECT
        Category,
        "Product ID",
        "Product Name",
        Total_Sales,
        Total_Profit,
        RANK() OVER (
            PARTITION BY Category
            ORDER BY Total_Profit DESC
        ) AS Profit_Rank
    FROM product_profit
)

SELECT *
FROM ranked_products
WHERE Profit_Rank <= 3
ORDER BY Category, Profit_Rank;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH product_profit AS (
    SELECT
        Category,
        "Product ID",
        "Product Name",
        SUM(Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM superstore
    GROUP BY
        Category,
        "Product ID",
        "Product Name"
),

ranked_products AS (
    SELECT
        Category,
        "Product ID",
        "Product Name",
        Total_Sales,
        Total_Profit,
        RANK() OVER (
            PARTITION BY Category
            ORDER BY Total_Profit ASC
        ) AS Loss_Rank
    FROM product_profit
)

SELECT *
FROM ranked_products
WHERE Loss_Rank <= 3
ORDER BY Category, Loss_Rank;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH yearly_sales AS (
    SELECT
        strftime('%Y', "Order Date") AS Year,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Year
)

SELECT
    Year,
    Total_Sales,
    LAG(Total_Sales) OVER (
        ORDER BY Year
    ) AS Previous_Year_Sales
FROM yearly_sales
ORDER BY Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH yearly_sales AS (
    SELECT
        strftime('%Y', "Order Date") AS Year,
        SUM(Sales) AS Total_Sales
    FROM superstore
    GROUP BY Year
),

sales_with_previous AS (
    SELECT
        Year,
        Total_Sales,
        LAG(Total_Sales) OVER (
            ORDER BY Year
        ) AS Previous_Year_Sales
    FROM yearly_sales
)

SELECT
    Year,
    Total_Sales,
    Previous_Year_Sales,
    ROUND(
        (Total_Sales - Previous_Year_Sales)
        * 100.0 / Previous_Year_Sales,
        2
    ) AS YoY_Growth_Percent
FROM sales_with_previous
ORDER BY Year;
"""

yoy_sales = pd.read_sql_query(query, conn)

yoy_sales
```

```python
query = """
WITH yearly_profit AS (
    SELECT
        strftime('%Y', "Order Date") AS Year,
        SUM(Profit) AS Total_Profit
    FROM superstore
    GROUP BY Year
),

profit_with_previous AS (
    SELECT
        Year,
        Total_Profit,
        LAG(Total_Profit) OVER (
            ORDER BY Year
        ) AS Previous_Year_Profit
    FROM yearly_profit
)

SELECT
    Year,
    Total_Profit,
    Previous_Year_Profit,
    ROUND(
        (Total_Profit - Previous_Year_Profit)
        * 100.0 / ABS(Previous_Year_Profit),
        2
    ) AS YoY_Profit_Growth_Percent
FROM profit_with_previous
ORDER BY Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
WITH customer_profit AS (
    SELECT
        "Customer ID",
        "Customer Name",
        Segment,
        SUM(Sales) AS Total_Sales,
        SUM(Profit) AS Total_Profit
    FROM superstore
    GROUP BY
        "Customer ID",
        "Customer Name",
        Segment
),

ranked_customers AS (
    SELECT
        "Customer ID",
        "Customer Name",
        Segment,
        Total_Sales,
        Total_Profit,

        RANK() OVER (
            ORDER BY Total_Profit DESC
        ) AS Profit_Rank

    FROM customer_profit
)

SELECT *
FROM ranked_customers
WHERE Profit_Rank <= 10
ORDER BY Profit_Rank;
"""

top_profit_customers = pd.read_sql_query(query, conn)

top_profit_customers
```

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    Segment,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY
    "Customer ID",
    "Customer Name",
    Segment
HAVING SUM(Profit) < 0
ORDER BY Total_Profit ASC;
"""

loss_customers = pd.read_sql_query(query, conn)

loss_customers.head(10)
```

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    Segment,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,

    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent

FROM superstore

GROUP BY
    "Customer ID",
    "Customer Name",
    Segment

ORDER BY Profit_Margin_Percent DESC;
"""

customer_margin = pd.read_sql_query(query, conn)

customer_margin.head(10)
```

```python
query = """
SELECT
    Segment,
    COUNT(DISTINCT "Customer ID") AS Total_Customers,
    COUNT(DISTINCT "Order ID") AS Total_Orders,
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,

    ROUND(
        SUM(Profit) * 100.0 / SUM(Sales),
        2
    ) AS Profit_Margin_Percent

FROM superstore

GROUP BY Segment

ORDER BY Total_Sales DESC;
"""

segment_customers = pd.read_sql_query(query, conn)

segment_customers
```
